```python
# Load train.csv
df1 = pd.read_csv("train.csv")
mriqc_columns = ["gsr_y", "dvars_vstd", "fd_mean", "fber", "tsnr", "snr"]
```

```python
exclusions=pd.read_csv("suggested_exclusions.csv")
exclusions["scan_id"]="sub-"+exclusions["scan_id"]
```

```python
exclusions = exclusions.loc[exclusions["MRIQC_fail"]==1.0,"scan_id"]
```

```python
df1.loc[df1["name"].isin(exclusions), "mriqc_fail"] = 1.0
```

```python
df1.to_csv("train.csv", index=False)
```

```python
means = df1[mriqc_columns].mean()
stds = df1[mriqc_columns].std()
```

```python
df1[mriqc_columns] = (df1[mriqc_columns] - means)/stds
```

```python
df1.to_csv("train_normalized_mriqc.csv", index=False)
```
